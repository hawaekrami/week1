## 3. Quick look at the data (2-3 real numbers)

Why these numbers matter: over half the dataset (54.2%) currently shows a
downward trend this is a common, recurring pattern, not a rare edge case,
so it's worth building a system around. Restricting to pages with real
traffic (>100 impressions/90d) still shows ~60% declining, meaning the
signal holds up even after filtering out low-volume noise. With 32 clients
and 30,000 pages, there's enough scale and diversity to look for patterns
that generalize, not just quirks of one site.

```python
import pandas as pd

df = pd.read_csv('/content_refresh_anonymized.csv')

# Number 1: base rate of decline
pct_declining = (df['trend_direction'] == 'down').mean() * 100
print(f"{pct_declining:.1f}% of pages are currently trending down")

# Number 2: decline rate among pages with real traffic (filtering out noise)
sizable = df[df['impressions_90d'] > 100]
pct_declining_sizable = (sizable['trend_direction'] == 'down').mean() * 100
print(f"Among pages with >100 impressions/90d ({len(sizable):,} pages), "
      f"{pct_declining_sizable:.1f}% are declining")

# Number 3: scale of the dataset
print(f"{df['client_id'].nunique()} clients, {len(df):,} content pages")
```

## 4. Careful words: what I can and can't claim

This work can say: observed decline rates, directional signals, and
decision-support rankings a prioritized list to help an editor decide
what to review first, backed by evidence from the data.

This work can never say: that a page's decline was CAUSED by anything
specific, or that refreshing a page WILL cause it to recover — proving
causation would require a real experiment (e.g. comparing refreshed vs.
un-refreshed pages over time), which is outside the scope of this data.

It also cannot claim to have discovered or predicted anything about
Google's actual ranking algorithm, or claim insight into AI search
rankings/citations the data only shows observed outcomes (impressions,
clicks, position, engagement), never the reasoning behind them.

Any score I build is a ranking tool for prioritization, not a guarantee.
A high-priority page might still not be worth fixing, and a low-priority
page might still matter — human judgment stays part of the process.
